# FIRST TRAIN

```python
from ultralytics import YOLO

# Load model
model = YOLO("models/yolo11s.pt")

# Train the model
results = model.train(
    data="datasets/Helmet-Detection-1/data.yaml",
    epochs=110,
    imgsz=416,
    batch=16,
    optimizer="Adam",
    lr0=0.001,
    plots=True,
    project="results/Hasil_Training_YOLOv11"
)
```

```python
# Load model hasil training
model = YOLO("results/Hasil_Training_YOLOv11/train/weights/best.pt")

# Validasi model
results = model.val(
    data="datasets/Helmet-Detection-1/data.yaml",
    save=True,
    project="results/runs/detect",
    name="val"
)
```
