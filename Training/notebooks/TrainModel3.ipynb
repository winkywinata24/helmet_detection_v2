# TRAINING

```python
from ultralytics import YOLO

# Load model
model = YOLO("models/yolo11l.pt")

# Train the model
results = model.train(
    data=yaml_path,
    epochs=150,
    imgsz=416,
    batch=16,
    optimizer='AdamW',
    lr0=1e-3,
    lrf=0.2,
    weight_decay=5e-4,
    augment=True,
    val=True,
    project="results/Hasil_Training_YOLOv11_3",
    exist_ok=True,
    patience=20,
    seed=42,
    device=0 
)
```

```python
# Load model hasil training
model = YOLO("results/Hasil_Training_YOLOv11_3/train/weights/best.pt")

# Validasi model
results = model.val(
    data="datasets/Helmet-Detection-14/data.yaml",
    save=True,
    project="results/runs3/detect",
    name="val"
)
```
